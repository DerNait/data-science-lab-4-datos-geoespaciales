## Hallazgos sólidos (bien respaldados por evidencia directa y consistente)

- Amatitlán tuvo cianobacteria sistemáticamente más alta que Atitlán en

- La floración de Amatitlán en 2026 fue un evento amplio (más de la mitad
  del lago con valores altos el 19 de junio de 2026) pero todavía no
  persistente en el tiempo (solo 0.15 % del área estuvo alta en la mitad o

  condición estable de todo el período.
- Atitlán se mantuvo consistentemente bajo en las cuatro fuentes de
  evidencia durante el período 2025-2026.

## Indicios (sugerentes, pero no concluyentes con estos datos)

- Un posible patrón estacional en Amatitlán (mayor valor en la única fecha
  de época lluviosa disponible) — se basa en una sola observación (`n=1`)
  y no se puede distinguir de la tendencia creciente general ya observada.
- La relación cianobacteria-NDVI más fuerte en Amatitlán que en Atitlán
  podría indicar que ahí la floración coincide más con biomasa/vegetación
  superficial detectable, pero no se midió composición de especies ni se
  confirmó en terreno.

## Limitaciones que impiden ir más allá

- Son 11 fechas irregulares por lago a lo largo de año y medio, no una
  serie continua: no alcanza para modelos de series de tiempo ni para
  afirmar estacionalidad robusta.
- Atitlán tuvo más fechas con advertencia de calidad de datos que
  Amatitlán en este período (ver ejercicio 7, sección 4); parte de su
  diferencia con Amatitlán podría estar influida por esa menor
  confiabilidad relativa, no solo por una diferencia biológica real.
- No se midieron directamente nutrientes, aguas residuales, temperatura
  del agua ni profundidad: las explicaciones de por qué Amatitlán muestra
  más cianobacteria son contexto documentado por las autoridades de cuenca
  de cada lago (ver ejercicio 7), no una causa demostrada por las imágenes
  satelitales de este laboratorio.
- El índice de cianobacteria es un proxy calibrado sobre datos simulados
  (ver `codebook.md`, ejercicio 3), no una medición de laboratorio in situ.

## Qué datos adicionales mejorarían el monitoreo

- Más fechas por lago, en especial durante la época lluviosa de
  Amatitlán, para poder evaluar la hipótesis estacional con más de una
  observación.
- Mediciones in situ de clorofila-a y conteo de cianobacterias para
  calibrar y validar el proxy satelital contra un dato de referencia real.
- Temperatura superficial del agua y datos meteorológicos (lluvia,
  viento) de estaciones cercanas a cada lago.
- Datos de nutrientes (nitrógeno, fósforo) y de descargas de aguas
  residuales en las cuencas, para poder relacionar directamente presión
  urbana con floraciones en vez de solo citarla como contexto.
